## 1. The Shell, Prompt, and Navigation

When you open a terminal, you are greeted by a **shell** (usually `bash` or `zsh`). The shell is a command-line interpreter that executes commands you type.

### Key Navigation Commands:
- `pwd` (**P**rint **W**orking **D**irectory): Shows the absolute path of the directory you are currently inside.
- `ls` (**L**i**s**t): Lists all files and folders in your current directory.
- `cd` (**C**hange **D**irectory): Moves you from one folder to another.

````python
# In a Jupyter notebook, you can run shell commands by prefixing them with '!'
print("--- Current Working Directory ---")
!pwd

print("\n--- Listing Files ---")
!ls -la
````

## 2. File & Directory Management

To organize your projects, data, and scripts, you need to know how to create, move, copy, and delete files and folders.

### Essential Commands:
- `mkdir folder_name`: Creates a new directory.
- `touch file_name.txt`: Creates an empty file (or updates its timestamp).
- `cp source destination`: Copies a file or directory.
- `mv source destination`: Moves or renames a file or directory.
- `rm file_name`: Deletes a file. (`rm -r` deletes a directory recursively).

````python
# Let's practice creating and managing a dummy workspace via the shell
!mkdir -p demo_workspace
!cd demo_workspace && touch data.csv notes.txt
!ls -la demo_workspace

# Clean up the demo workspace
!rm -rf demo_workspace
print("\nDemo workspace cleaned up!")
!ls -la demo_workspace
````

## 3. Permissions, Ownership, and `sudo`

Linux is a multi-user operating system, meaning files are protected by strict permissions defining who can **R**ead, **W**rite, or e**X**ecute them.

### Managing Permissions:
- `ls -l`: Displays file permissions (e.g., `-rwxr-xr--`). The first character indicates file type (`-` for file, `d` for directory), followed by three sets of permissions for the **Owner**, **Group**, and **Others**.
- `chmod`: Changes file permissions (e.g., `chmod +x script.sh` makes a script executable).
- `chown`: Changes file ownership.

### The Power of `sudo`:
- **`sudo`** (**S**uper**U**ser **Do**): Executes a command with administrative (root) privileges. Use it carefully when installing packages (e.g., `sudo apt install htop`) or modifying system configurations.

___

## 4. Viewing and Searching Text Files

In science and development, you will spend a lot of time inspecting configuration files, logs, or tabular data outputs directly from the terminal.

### Quick Inspection Tools:
- `cat filename`: Prints the entire contents of a file to the screen (best for small files).
- `less filename`: Opens an interactive, scrollable viewer for large files (press `q` to exit).
- `head -n 5 filename`: Prints the first 5 lines of a file.
- `tail -n 5 filename`: Prints the last 5 lines of a file (great for monitoring live logs using `tail -f`).

### Searching Text:
- `grep "keyword" filename`: Searches for matching lines containing a specific string or regular expression inside a file.

## 5. Remote Access: SSH & SFTP

When working with institutional clusters, cloud instances, or remote telescopes, your local laptop won't have enough muscle. You must connect remotely.

### 1. Secure Shell (`ssh`)
`ssh` allows you to log into a remote machine securely over an encrypted connection.
- **Basic Usage:** `ssh username@hostname_or_ip`
- **Using an Identity Key:** `ssh -i ~/.ssh/id_rsa username@server.edu`

### 2. Secure File Transfer Protocol (`sftp`)
If you need to download a large dataset or upload scripts to a remote server without using Git:
- **Connect:** `sftp username@hostname_or_ip`
- **Commands inside SFTP:** 
  - `get remote_file.fits` (downloads file to your local computer)
  - `put local_script.py` (uploads file to the remote server)
  - `bye` (closes the session)

## 6. Customizing Your Shell (`.bashrc`)

Every time you open a terminal, your shell executes a configuration script—usually `~/.bashrc` (or `~/.zshrc` if you use Zsh). 

You can customize this file to:
1. **Define Environment Variables:** Store API keys, paths, or default configurations.
2. **Create Aliases:** Shortcuts for long, repetitive commands.

### Example Customizations to add to your `~/.bashrc`:
```bash
# Useful shortcuts (Aliases)
alias ll='ls -la'
alias cls='clear'
alias python='python3'

# Custom Environment Variables
export ASTRO_DATA_DIR="/home/user/data/fits_files/"
```

After editing your `.bashrc`, remember to apply changes instantly by running:
```bash
source ~/.bashrc
```
